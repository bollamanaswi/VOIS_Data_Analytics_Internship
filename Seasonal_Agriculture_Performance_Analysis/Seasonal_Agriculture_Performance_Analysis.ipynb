```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('/content/seasonal_agriculture_performance_dataset.csv')

# Display first 5 rows
df.head()
```

```python
#Check missing values
# Check missing values
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)

# Total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Fill missing numerical values with median
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
#Yield — cap the outliers
# Remove/Capping Yield Outliers

Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].clip(
    lower=lower_limit,
    upper=upper_limit
)

print("Yield outliers capped successfully.")
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
#Statistical significance — ANOVA
from scipy.stats import f_oneway

season_groups = [
    group['Yield_Tonnes_Ha'].dropna()
    for name, group in df.groupby('Season')
]

f_stat, p_value = f_oneway(*season_groups)

print("ANOVA Results")
print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Yield differs significantly across seasons.")
else:
    print("Result: No statistically significant difference in yield across seasons.")
```

```python
profit_groups = [
    group['Profit_INR'].dropna()
    for name, group in df.groupby('Season')
]

f_stat, p_value = f_oneway(*profit_groups)

print("ANOVA for Profit")
print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Profit differs significantly across seasons.")
else:
    print("Result: No statistically significant difference in profit across seasons.")
```
