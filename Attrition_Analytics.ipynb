```python
# Define a list of important columns to be used for further analysis
# This selects a subset of the original DataFrame to focus on relevant features.
selected_cols=['EmpID','Age','AgeGroup','Department','Gender','EducationField','SalarySlab','YearsAtCompany','MonthlyIncome','Attrition']
```

```python
# Create a new DataFrame 'df1' containing only the selected columns.
# Use .copy() to ensure df1 is an independent copy and prevent SettingWithCopyWarning.
df1=df[selected_cols].copy()
# Create a new numerical column 'Attrition_Numeric' from the 'Attrition' column.
# 'Yes' is mapped to 1 (indicating attrition) and 'No' is mapped to 0 (indicating no attrition).
# This conversion is crucial for numerical analysis and plotting of attrition rates.
df1['Attrition_Numeric'] = df1['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)
# Display the first few rows of the new DataFrame to verify the changes.
df1.head()
```

```python
# Define a function to identify and return the lower and upper bounds for outlier removal
# using the Interquartile Range (IQR) method. Outliers are typically considered values
# that fall below Q1 - 1.5*IQR or above Q3 + 1.5*IQR.
def remove_outlier(col_name):
  sorted(col_name)
  # Calculate the first quartile (Q1) and third quartile (Q3)
  Q1,Q3=col_name.quantile([0.25,0.75])
  # Calculate the Interquartile Range (IQR)
  IQR=Q3-Q1
  # Calculate the lower bound for outliers
  lower_range=Q1-(1.5*IQR)
  # Calculate the upper bound for outliers
  upper_range=Q3+(1.5*IQR)
  return lower_range,upper_range
```

```python
low, high = remove_outlier(df1['YearsAtCompany'])
```

```python
df1['YearsAtCompany'] = np.where(df1['YearsAtCompany']>high,high,df1['YearsAtCompany'])
```

```python
df1['YearsAtCompany'] = np.where(df1['YearsAtCompany']<low,low,df1['YearsAtCompany'])
```

```python
sns.boxplot(y='YearsAtCompany',data=df1)
```

```python
low , high = remove_outlier(df1['MonthlyIncome'])
```

```python
df1['MonthlyIncome'] = np.where(df1['MonthlyIncome']>high,high,df1['MonthlyIncome'])
```

```python
df1['MonthlyIncome'] = np.where(df1['MonthlyIncome']<low,low,df1['MonthlyIncome'])
```
